# **Section 3: Advanced data Exploration and feature selection**

In this notebook, we will perform advanced data cleaning and feature selection to reduce the dimensionality of our data set.

We will decide which features to keep based on the **training data** only to prevent data leakage. The selected features are then applied to test data.

### **Contents:**
#### **1. Load the data**

#### **2. Outlier removal**
We will then perform outlier removal to remove extreme samples from the data set. As we saw with our preliminary data analysis, some of our samples have extreme values, which in our case comes from bad data quality. We will identify these samples based on thresholds defined using the training set, and remove the samples above this threshold in both training and test sets. 

#### **3. Remove constant features**
We will then remove features with very little variance as they do not provide information useful for a machine learning classification problem. 

#### **4. Data exploration**
We will perform some simple data exploration and visualize what our training and test data sets look like after removing outliers and constant features.

#### **5. Select top features using SelectKBest**
We will then use **SelectKBest** from scikitlearn to keep only the most discriminative features. I will use an ANOVA F-score and keep only significant features.

#### **6. Aggregate similar features**
Finally, we will reduce the dimensionality of the feature matrix by grouping together features that are very similar or identical. To do this, we will use the function **FeatureAgglomeration** from scikitlearn. This function groups similar features and combines them into a new feature that contains the average value for all the features in that group. We will save which features were grouped together for further analysis.

#### **7. Save selected feature matrix**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Set styles
sns.set_style('whitegrid')
sns.set_palette('pastel')
```

## **1. Load Data**

```python
# Load training features and labels
cnv_train = pd.read_csv('Data/cnv_train.csv', index_col=0)
cnv_test = pd.read_csv("Data/cnv_test.csv", index_col=0)
sxt_train = pd.read_csv('Data/SXT_train.csv', index_col=0)
sxt_test = pd.read_csv("Data/SXT_test.csv", index_col=0)

print(f'Training set shape: {cnv_train.shape}')
print(f'Labels shape: {sxt_train.shape}')
print(f'Resistance distribution:')
print(sxt_train.value_counts())
```

## **2. Outlier removal**

We will first remove outlier samples from the analysis. As we saw in notebook 1, several samples had overall values that were above average or that had a higher than average standard deviation. 
We will identify these samples using thresholds calculated on the training set and remove them from the training and test sets. 

### **2.1 Identify Outliers**
We will flag and remove samples whose average CNV or standard deviation deviate by more than 2.5 standard deviations from the population distribution. This will thus remove extreme values from the dataset.

We will define the thresholds based on the distribution of the **training** set, and then remove samples above the thresholds from both the **training** and **test** set. We took the decision to remove these outliers from the training and test set, as these samples are extreme do to bad quality data. 

```python
# Calculate statistics
stats_train = pd.DataFrame({
    'mean': cnv_train.mean(axis=1),
    'std': cnv_train.std(axis=1)
})

# Define outlier thresholds
# Mean
mean_upper = stats_train['mean'].mean() + 2.5 * stats_train['mean'].std()
mean_lower = stats_train['mean'].mean() - 2.5 * stats_train['mean'].std()

# Std
std_upper = stats_train['std'].mean() + 2.5 * stats_train['std'].std()
std_lower = stats_train['std'].mean() - 2.5 * stats_train['std'].std()

# Identify outliers
outlier_mean = (stats_train['mean'] > mean_upper) | (stats_train['mean'] < mean_lower)
outlier_std = (stats_train['std'] > std_upper) | (stats_train['std'] < std_lower)

print(f'outliers based on mean: {outlier_mean.sum()}')
print(f'outliers based on std: {outlier_std.sum()}')

# Combine all outliers
all_outliers = outlier_mean | outlier_std

print(f'Total outliers: {all_outliers.sum()}')
```

### **2.2 Remove Outliers from training set**

```python
print(f'Before outlier removal: {cnv_train.shape[0]} samples')

# Remove outliers
cnv_train = cnv_train[~all_outliers]
sxt_train = sxt_train[~all_outliers]

print(f'After outlier removal: {cnv_train.shape[0]} samples')
print(f'Removed: {all_outliers.sum()} samples')
print(f'\nUpdated resistance distribution:')
print(sxt_train.value_counts())
```

### **2.3 Remove Outliers from test set**
Let's now use the same thresholds to remove outliers from the test set.

```python
# Calculate statistics for test set
stats_test = pd.DataFrame({
    'mean': cnv_test.mean(axis=1),
    'std': cnv_test.std(axis=1)
})

# Identify outliers
outlier_mean_test = (stats_test['mean'] > mean_upper) | (stats_test['mean'] < mean_lower)
outlier_std_test = (stats_test['std'] > std_upper) | (stats_test['std'] < std_lower)

print(f'outliers based on mean: {outlier_mean_test.sum()}')
print(f'outliers based on std: {outlier_std_test.sum()}')

# Combine all outliers
all_outliers_test = outlier_mean_test | outlier_std_test

print(f'Total outliers: {all_outliers_test.sum()}')
```

```python
print(f'Before outlier removal: {cnv_test.shape[0]} samples')

# Remove outliers
cnv_test = cnv_test[~all_outliers_test]
sxt_test = sxt_test[~all_outliers_test]

print(f'After outlier removal: {cnv_test.shape[0]} samples')
print(f'Removed: {all_outliers_test.sum()} samples')
print(f'\nUpdated resistance distribution:')
print(sxt_test.value_counts())
```

## **3. Remove constant features**
Let's now remove features that have very little variance (that are almost the same for all samples). As before, let's choose which features to remove on the training set and then remove the same features from the test set.

To calculate the variance, I will compute the standard deviation for all features and remove all features that have a standard deviation below 0.01 in the training set. 

### **3.1 Visualizing the std for all features**
Let's first visualize the std for all of our features using a histogram.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

sns.histplot(cnv_train.std(axis=0), bins=100, log_scale=True, color='lightblue', edgecolor='black', ax=axes[0])
axes[0].set_title("Std per feature")
axes[0].set_xlabel("Std (log scale)")
axes[0].set_ylabel("Number of features")
axes[0].set_xscale("log")

sns.histplot(cnv_train.std(axis=0), bins=100, log_scale=True, color='lightblue', edgecolor='black', ax=axes[1], cumulative=True, stat="percent")
axes[1].set_title("Cumulative Std")
axes[1].set_xlabel("Std (log scale)")
axes[1].set_ylabel("% of features")
axes[1].set_xscale("log")

plt.tight_layout()
plt.savefig('Figures/03_Std.png', dpi=300, bbox_inches='tight')
plt.show()
```

We see that the variance for each feature follows a standard deviation with a tail to the right. 
Based on this graph, we will remove all features that have a standard deviation below 0.1 which represents around 10% of features. These likely contribute little to explaining the differences between the sample

```python
low_variance_mask = cnv_train.std(axis=0) < 0.1
print(f'{low_variance_mask.sum()} features have low variance')
```

```python
print(f'Before removing features: {cnv_train.shape}')
cnv_train = cnv_train.loc[:,~low_variance_mask]

print(f'After removing features: {cnv_train.shape}')
```

```python
print(f'Before removing features: {cnv_test.shape}')
cnv_test = cnv_test.loc[:, ~low_variance_mask]
print(f'After removing features: {cnv_test.shape}')
```

### **4.5 Volcano plot (training set)**

Finally, I will redo the volcano plot that I did in the first notebook to explore how many features are significantly different between the two categories (resistant vs susceptible) **in the training set**. As explained in notebook 01, I will see if each feature is significantly different between the resistant and susceptible strains using a Mann-Whitney U test followed by Benjamin-Hochberg FDR correction (which corrects for the high number of samples). I will also calculate the log2 fold change for each features, and visualize the results. Let's do this visualization for the training set only.


```python
from scipy.stats import mannwhitneyu, false_discovery_control

resistant = cnv_train.loc[sxt_train.iloc[:, 0] == 1]
susceptible = cnv_train.loc[sxt_train.iloc[:, 0] == 0]

genes = cnv_train.columns
```

```python
U, p = mannwhitneyu(resistant, susceptible, axis=0, alternative="two-sided")
mask = ~np.isnan(p)  
p = p[mask]
genes = genes[mask]
padj = false_discovery_control(p, method="bh")
nlog10padj = -np.log10(np.clip(padj, np.finfo(float).tiny, None))
log2FC = (susceptible.mean() - resistant.mean()).values[mask]
```

```python
up = (padj < 0.01) & (log2FC >  0.3)   # higher in susceptible
down = (padj < 0.01) & (log2FC < -0.3)   # higher in resistant
ns = ~(up | down)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(log2FC[ns], nlog10padj[ns], s=8, c="lightgrey", alpha=0.6)
ax.scatter(log2FC[up], nlog10padj[up], s=12, c="salmon")
ax.scatter(log2FC[down], nlog10padj[down], s=12, c="lightblue")

ax.axvline(0.3, ls="--", c="grey", lw=1)
ax.axvline(-0.3, ls="--", c="grey", lw=1)
ax.axhline(-np.log10(0.01), ls="--", c="grey", lw=1)

ax.set_xlabel("log2 fold change (susceptible vs resistant)")
ax.set_ylabel("-log10(adjusted p-value)")
ax.set_title("Copy number differences between susceptible and resistant samples")

ax.legend(frameon=False, loc="upper left")
plt.tight_layout()
plt.savefig('Figures/03_Volcano_train.png', dpi=300, bbox_inches='tight')
plt.show()
```

Interestingly, even after removing outliers and performing feature selection, the data is still skewed in one direction, with a lot of genes being significantly more present in **resistant** bacteria than in the susceptible ones. This could be explained by some genes not present at all in most of the susceptible strains, but on average more present in the resistant strain. This is biologically interesting: resistant burkholderia are known to have higher copies of specific genomic regions. Let's quickly calculate a score for each feature and list the 20 regions with the highest score to see if they are known regions linked to resistance.

```python
score = np.abs(log2FC) * -np.log10(padj)
score_df = pd.DataFrame({
    "gene": genes, 
    "score": score
})
```

```python
score_df.sort_values("score", ascending=True)["gene"].head(20)
```

When looking at this list, some candidates stand out such as the tetR gene. TetR are transcriptional regulators that are commonly associated with antibiotic resistance. 
It's important to note that seeing genes that are high on the list here does not necessarily mean that there is a causal mechanism. Indeed, they could simply be genes that are indirectly more highly present in resistant bacteria (such as genes linked to metabolism or cell cycle for instance). This could very well be the case for the regions annotated is Glutathione, Nitrite or Acyl-CoA. 
It will be interesting to see if a model uses the same feature to predict resistance!

## **5. Select Features using SelectKBest**

I will now select features using SelectKBest and an ANOVA F-score. This identifies the most discriminative features. I will choose to keep only features that are significantly associated with the resistance profile of our samples.

### **5.1 Score all features**

```python
#scale features
scaler = StandardScaler().fit(cnv_train)
cnv_train_scaled = scaler.transform(cnv_train)
cnv_test_scaled = scaler.transform(cnv_test)
```

```python
from scipy.stats import false_discovery_control

# Calculate F-scores for all features
selector = SelectKBest(score_func=f_classif, k='all')
selector.fit(cnv_train_scaled, sxt_train.iloc[:, 0])

feature_importance = pd.DataFrame({
    'feature': cnv_train.columns,
    'f_score': selector.scores_,
    'p_value': selector.pvalues_,
})

feature_importance['p_adj'] = false_discovery_control(feature_importance['p_value'], method='bh')
feature_importance = feature_importance.sort_values('f_score', ascending=False)

significant = feature_importance['p_adj'] < 0.05
print(f"Significant features: {significant.sum()} ({significant.mean() * 100:.1f}%)")
```

### **5.2 Select Significant Features (p < 0.05)**
Let's now select only significant features.

```python
# Select all significant features
sig_features = feature_importance.loc[feature_importance['p_adj'] < 0.05, 'feature']
cnv_train_selected = cnv_train[sig_features]
cnv_test_selected = cnv_test[sig_features]

print(f'Features before selection: {cnv_train.shape[1]}')
print(f'Features after selection: {cnv_train_selected.shape[1]}')
```

### **5.3 Statistics per feature**

```python
# Calculate per-sample statistics
sample_stats_selected = pd.DataFrame({
    'sample': cnv_train_selected.columns,
    'mean': cnv_train_selected.mean(axis=0).values,
    'median': cnv_train_selected.median(axis=0).values,
    'std': cnv_train_selected.std(axis=0).values,
    'min': cnv_train_selected.min(axis=0).values,
    'max': cnv_train_selected.max(axis=0).values
})
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Mean distribution
axes[0, 0].hist(sample_stats_selected['median'], bins=50, alpha=0.7, edgecolor='black')
axes[0, 0].set_xlabel('Median CNV (log2 ratio)', fontweight='bold')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title('Distribution of Median CNV per region', fontweight='bold')
axes[0, 0].grid(alpha=0.3)

# Standard deviation distribution
axes[0, 1].hist(sample_stats_selected['std'], bins=50, alpha=0.7, edgecolor='black')
axes[0, 1].set_xlabel('CNV Standard Deviation', fontweight='bold')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].set_title('Distribution of CNV Std per region', fontweight='bold')
axes[0, 1].grid(alpha=0.3)

# Max value distribution
axes[1, 0].hist(sample_stats_selected['max'], bins=50, alpha=0.7, edgecolor='black')
axes[1, 0].set_xlabel('Max CNV (log2 ratio)', fontweight='bold')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title('Max CNV per region', fontweight='bold')
axes[1, 0].grid(alpha=0.3)

# Min value distribution
axes[1, 1].hist(sample_stats_selected['min'], bins=50, alpha=0.7, edgecolor='black')
axes[1, 1].set_xlabel('Min CNV (log2 ratio)', fontweight='bold')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_title('Min CNV per region', fontweight='bold')
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('Figures/03_statistics_selected_train.png', dpi=300, bbox_inches='tight')
plt.show()
```

### **5.4 Volcano plot**
Let's redo the same volcano plot as before with these features. 

```python
resistant_selected = cnv_train_selected.loc[sxt_train.iloc[:, 0] == 1]
susceptible_selected = cnv_train_selected.loc[sxt_train.iloc[:, 0] == 0]

genes_selected = cnv_train_selected.columns
```

```python
U, p = mannwhitneyu(resistant_selected, susceptible_selected, axis=0, alternative="two-sided")
mask = ~np.isnan(p)  
p = p[mask]
genes_selected = genes_selected[mask]
padj = false_discovery_control(p, method="bh")
nlog10padj = -np.log10(np.clip(padj, np.finfo(float).tiny, None))
log2FC = (susceptible_selected.mean() - resistant_selected.mean()).values[mask]
```

```python
up = (padj < 0.01) & (log2FC >  0.3)   # higher in susceptible
down = (padj < 0.01) & (log2FC < -0.3)   # higher in resistant
ns = ~(up | down)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(log2FC[ns], nlog10padj[ns], s=8, c="lightgrey", alpha=0.6)
ax.scatter(log2FC[up], nlog10padj[up], s=12, c="salmon")
ax.scatter(log2FC[down], nlog10padj[down], s=12, c="lightblue")

ax.axvline(0.3, ls="--", c="grey", lw=1)
ax.axvline(-0.3, ls="--", c="grey", lw=1)
ax.axhline(-np.log10(0.01), ls="--", c="grey", lw=1)

ax.set_xlabel("log2 fold change (susceptible vs resistant)")
ax.set_ylabel("-log10(adjusted p-value)")
ax.set_title("Copy number differences between susceptible and resistant samples")
ax.legend(frameon=False, loc="upper left")
ax.legend(frameon=False, loc="upper left")

plt.tight_layout()
plt.savefig('Figures/03_Volcano_selected_train.png', dpi=300, bbox_inches='tight')
plt.show()
```

It looks like most genes that had a high p-value and a low log2 fold change were removed, as expected. Interestingly, we now have much less genes that are present in more copies in the susceptible strains than before. The distribution all together still seems skewed, with a significant amount of genes present in more copies in the resistant strains than in the susceptible strains. 

## **6. Dimensionality Reduction**
### **6.1 Aggreagte similar features**
Let's now reduce the dimensionality of the feature matrix by grouping together features that are very similar. To do so, I will use the function **FeatureAgglomeration** from scikitlearn.

This functions groups similar features and combines them into a new feature that contains the average value for all the features in that group. We will group together features that have a correlation between each other >0.90

```python
from sklearn.cluster import FeatureAgglomeration
# Cluster features
agglo = FeatureAgglomeration(n_clusters=None, linkage="average", metric="correlation", distance_threshold=0.1, pooling_func=np.median,)
agglo.fit(cnv_train_selected)

print(f'We go from {cnv_train_selected.shape[1]} features to {agglo.n_clusters_} features')
```

```python
# Cluster features
cnv_train_reduced = agglo.transform(cnv_train_selected)
cnv_test_reduced = agglo.transform(cnv_test_selected)

# back to df
col_names = range(agglo.n_clusters_)
cnv_train_reduced = pd.DataFrame(cnv_train_reduced, columns=col_names, index=cnv_train.index)
cnv_test_reduced = pd.DataFrame(cnv_test_reduced, columns=col_names, index=cnv_test.index)
```

```python
cnv_train_reduced.shape
```

Let's also save which features were grouped together for further analysis

```python
feature_names = cnv_train_selected.columns 
groups = pd.DataFrame({
    'feature': feature_names,
    'cluster': agglo.labels_
})

clusters = groups.groupby("cluster")["feature"].apply(list)
print(f'{(clusters.apply(len) > 1).sum()} clusters were made from more than one feature')
```

```python
groups.head()
```

```python
groups.to_csv("Data/clusters.csv", index=False)
```
