# **Section 3: Advanced data cleaning and feature selection**

In this notebook, we will perform advanced data cleaning and feature selection to reduce the dimensionality of our data set.

We will decide which features to keep based on the **training data** only to prevent data leakage. The selected features are then applied to test data.

### **Contents:**
#### **1. Load the data**

#### **2. Outlier removal**
We will first perform outlier removal to remove extreme samples from the data set. As we saw with our preliminary data analysis, some of our samples have extreme values, which in our case comes from bad data quality. We will identify these samples based on thresholds defined using the training set, and remove the samples above this threshold in both training and test sets. 

#### **3. Remove constant features**
We will then remove features with very little variance as they do not provide information useful for a machine learning classification problem. 

#### **3. Aggregate similar features**

#### **4. Select top features using SelectKBest**

#### **5. Save selected feature matrix**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

# Set styles
sns.set_style('whitegrid')
sns.set_palette('pastel')
```

## **1. Load Data**

```python
# Load training features and labels
cnv_train = pd.read_csv('Data/cnv_train.csv', index_col=0)
cnv_test = pd.read_csv("Data/cnv_test.csv", index_col=0)
sxt_train = pd.read_csv('Data/SXT_train.csv', index_col=0)
sxt_test = pd.read_csv("Data/SXT_test.csv", index_col=0)

print(f'Training set shape: {cnv_train.shape}')
print(f'Labels shape: {sxt_train.shape}')
print(f'Resistance distribution:')
print(sxt_train.value_counts())
```

## **2. Outlier removal**

Remove outlier samples from the analysis. As we saw in notebook 1, several samples had overall values that were above average or a higher than average standard deviation. 
We will identify these samples using thresholds calculated on the training set and remove them from the training and test sets. 

### **2.1 Identify Outliers**
We will remove samples whose mean or standard deviation deviate by more than 2.5 standard deviations from the population distribution. This will thus remove extreme values from the dataset.

We will define the thresholds based on the distribution of the **training** set, and then remove samples above the thresholds from both the **training** and **test** set. We took the decision to remove these outliers from the training and test set, as these samples are extreme do to bad quality data. 

```python
# Calculate statistics
stats_train = pd.DataFrame({
    'mean': cnv_train.mean(axis=1),
    'std': cnv_train.std(axis=1),
    'max': cnv_train.max(axis=1)
})

# Define outlier thresholds
# Mean
mean_upper = stats_train['mean'].mean() + 2.5 * stats_train['mean'].std()
mean_lower = stats_train['mean'].mean() - 2.5 * stats_train['mean'].std()

# Std
std_upper = stats_train['std'].mean() + 2.5 * stats_train['std'].std()
std_lower = stats_train['std'].mean() - 2.5 * stats_train['std'].std()

# Identify outliers
outlier_mean = (stats_train['mean'] > mean_upper) | (stats_train['mean'] < mean_lower)
outlier_std = (stats_train['std'] > std_upper) | (stats_train['std'] < std_lower)

print(f'outliers based on mean: {outlier_mean.sum()}')
print(f'outliers based on std: {outlier_std.sum()}')

# Combine all outliers
all_outliers = outlier_mean | outlier_std

print(f'Total outliers: {all_outliers.sum()}')
```

### **2.2 Remove Outliers from training set**

```python
print(f'Before outlier removal: {cnv_train.shape[0]} samples')

# Remove outliers
cnv_train = cnv_train[~all_outliers]
sxt_train = sxt_train[~all_outliers]

print(f'After outlier removal: {cnv_train.shape[0]} samples')
print(f'Removed: {all_outliers.sum()} samples')
print(f'\nUpdated resistance distribution:')
print(sxt_train.value_counts())
```

Let's now use the same thresholds to remove outliers from the test set.

## **3. Remove constant features**
Let's now remove features that have very little variance (that are almost the same for all of the samples). As before, let's choose which features to remove on the training set and then remove the same features from the test set.

To calculate the variance, I will compute the standard deviation for all features and remove all features that have a standard deviation below 0.01 in the training set. 

### **3.1 Visualizing the std for all features**
Let's first visualize the std for all of our features using a histogram.

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

sns.histplot(cnv_train.std(axis=0), bins=100, log_scale=True, color='lightblue', edgecolor='black', ax=axes[0])
axes[0].set_title("Std per feature")
axes[0].set_xlabel("Std (log scale)")
axes[0].set_ylabel("Number of features")
axes[0].set_xscale("log")

sns.histplot(cnv_train.std(axis=0), bins=100, log_scale=True, color='lightblue', edgecolor='black', ax=axes[1], cumulative=True, stat="percent")
axes[1].set_title("Cumulative Std")
axes[1].set_xlabel("Std (log scale)")
axes[1].set_ylabel("% of features")
axes[1].set_xscale("log")

plt.tight_layout()
plt.show()
```

We see that the variance for each feature follows a standard deviation with a tail to the right. 
Based on this graph, we will remove all features that have a standard deviation below 0.1 which represents around 10% of features. These likely contribute little to explaining the differences between the sample

```python
low_variance_mask = cnv_train.std(axis=0) < 0.1
print(f'{low_variance_mask.sum()} features have low variance')
```

```python
print(f'Before removing features: {cnv_train.shape}')
cnv_train = cnv_train.loc[:,~low_variance_mask]

print(f'After removing features: {cnv_train.shape}')
```

```python
print(f'Before removing features: {cnv_test.shape}')
cnv_test = cnv_test.loc[:, ~low_variance_mask]
print(f'After removing features: {cnv_test.shape}')
```

## **4. Aggreagte similar features**
Let's now reduce the dimensionality of the feature matrix by grouping together features that are very similar. To do so, I will use the function **FeatureAgglomeration** from scikitlearn.

This functions groups similar features and combines them into a new feature that contains the average value for all the features in that group. We will group together features that have a correlation between each other >0.95

It is important to scale the data beforehand, since a feature with high values can skew the group average when grouping the features together. 

```python
from sklearn.cluster import FeatureAgglomeration

# scale features
cnv_train_scaled = StandardScaler().fit_transform(cnv_train)
cnv_test_scaled = StandardScaler().fit_transform(cnv_test)

# Cluster features
agglo = FeatureAgglomeration(n_clusters=None, linkage="average", metric="correlation", distance_threshold=0.05)
agglo.fit(cnv_train_scaled)

print(f'We go from {cnv_train.shape[1]} features to {agglo.n_clusters_} features')
```

```python
# Cluster features
cnv_train_reduced = agglo.transform(cnv_train_scaled)
cnv_test_reduced = agglo.transform(cnv_test_scaled)

# back to df
col_names = range(agglo.n_clusters_)
cnv_train_reduced = pd.DataFrame(cnv_train_reduced, columns=col_names, index=cnv_train.index)
cnv_test_reduced = pd.DataFrame(cnv_test_reduced, columns=col_names, index=cnv_test.index)
```

Let's also save which features were grouped together for further analysis

```python
feature_names = cnv_train.columns 
groups = pd.DataFrame({
    'feature': feature_names,
    'cluster': agglo.labels_
})

clusters = groups.groupby("cluster")["feature"].apply(list)
print(f'{(clusters.apply(len) > 1).sum()} clusters were made from more than one feature')
```

```python
groups.to_csv("Data/clusters.csv", index=False)
```

## **5. Select Features using SelectKBest**

I will now select features using SelectKBest and an ANOVA F-score. This identifies the most discriminative features. I will choose to keep only features that are significantly associated with the resistance profile of our samples.

### **5.1 Score all features**

```python
# Calculate F-scores for all features
selector = SelectKBest(score_func=f_classif, k='all')
selector.fit(cnv_train_reduced.values, sxt_train.iloc[:, 0])

# Get scores for all features
feature_scores = selector.scores_
feature_pvalues = selector.pvalues_

# Create results dataframe
feature_importance = pd.DataFrame({
    'feature': cnv_train_reduced.columns,
    'f_score': feature_scores,
    'p_value': feature_pvalues
})

feature_importance = feature_importance.sort_values('f_score', ascending=False)
feature_importance = feature_importance.dropna()

print(f'Significant features (p<0.05): {(feature_pvalues < 0.05).sum()} ({((feature_pvalues < 0.05).sum())/len(feature_pvalues)*100:.1f})%')
```

### **5.2 Select Significant Features (p < 0.05)**

```python
# Select all significant features
sig_mask = (feature_pvalues < 0.05)
cnv_train_selected = cnv_train_reduced.loc[:, sig_mask]
cnv_test_selected = cnv_test_reduced.loc[:, sig_mask]

print(f'Features before selection: {cnv_train_reduced.shape[1]}')
print(f'Features after selection: {cnv_train_selected.shape[1]}')
```

### **4.3 Visualize Feature Selection Results**

```python
# Build cluster -> list of original feature names
cluster_to_features = groups.groupby('cluster')['feature'].apply(list)

# Remove NaN scores for visualization
feature_importance_clean = feature_importance.dropna()
fig, axes = plt.subplots(1, 4, figsize=(26, 6))

# Distribution of F-scores
ax = axes[0]
ax.hist(feature_importance_clean['f_score'], bins=50, alpha=0.7, edgecolor='black')
ax.axvline(feature_importance_clean['f_score'].quantile(0.95), 
           color='red', linestyle='--', linewidth=2, label='95th percentile')
ax.set_xlabel('F-Score (ANOVA)', fontweight='bold')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Feature F-Scores', fontweight='bold')
ax.legend()
ax.grid(alpha=0.3)

# Elbow plot
cumsum_scores = np.cumsum(feature_importance_clean['f_score'].values)
cumsum_scores_pct = 100 * cumsum_scores / cumsum_scores[-1]
ax = axes[1]
ax.plot(range(1, len(cumsum_scores_pct)+1), cumsum_scores_pct, linewidth=2)
ax.axhline(95, color='red', linestyle='--', linewidth=2, label='95% cumulative score')
ax.set_xlabel('Number of Features', fontweight='bold')
ax.set_ylabel('Cumulative % of F-Score', fontweight='bold')
ax.set_title('Elbow Plot: Feature Importance Accumulation', fontweight='bold')
ax.legend()
ax.grid(alpha=0.3)

# Top 20 features
ax = axes[2]
top20 = feature_importance_clean.head(20).sort_values('f_score')
bars = ax.barh(range(len(top20)), top20['f_score'], edgecolor='black', linewidth=1)
ax.set_yticks(range(len(top20)))
ax.set_yticklabels(top20["feature"], fontsize=9)
ax.set_xlabel('F-Score (ANOVA)', fontweight='bold')
ax.set_title('Top 20 Most Discriminative Features', fontweight='bold')
ax.grid(axis='x', alpha=0.3)

# Feature ID to original feature name(s)
ax = axes[3]
ax.axis('off')
lines = []
for _, row in top20.iterrows():
    feature_id = int(row['feature'])
    original_features = clusters.get(feature_id, [])
    original_string = ', '.join(original_features)
    lines.append(f"{feature_id}: {original_string}")
ax.text(0, 1, '\n'.join(lines), fontsize=10, va='top')
ax.set_title('Feature ID to Original Feature(s)', fontweight='bold')

plt.tight_layout()
plt.savefig('Figures/04_feature_selection.png', dpi=300, bbox_inches='tight')
plt.show()
```
