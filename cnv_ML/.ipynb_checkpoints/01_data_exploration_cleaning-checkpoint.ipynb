# **Section 1: CNV Data Exploration, Cleaning, and Preprocessing**

The data for this project consists of a metadata file with information about the samples (including their resistance profile), and a cnv_depth_matrix.tsv file prepared using the Snakemake Pipeline. The cnv_depth_matrix.tsv file contains the copy number variations in genes and promoters for each sample.

In this notebook, information about the antibiotic we are focusing on, **SXT** is parsed from the metadata file. Each sample is then classified as either *Resistant*, *Susceptible* or *NaN* (not tested) depending on the information in the metadata file. The CNV feature matrix is then loaded and preprocessed to ensure that all samples present in the file are classified as either *Resistant* or *Susceptible*. Samples that do not have a resistance profile for SXT are removed from the analysis.

Preliminary data exploration is performed and the data is cleaned to remove errors or empty values.

This notebook contains the following sections:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style('whitegrid')
sns.set_palette('pastel')
```

## **0. Helper Function**
We will first write a helper function that will help us parse the resistance profile information present in the metadata file. 

The function **parse_resistance_profile** extracts resistance information from genome_metadata.csv. It takes the string present in the *resistance_profile* column of the metadata file as input. 

The string containing the resistance information is structured as follows:

"Resistant to SXT; Susceptible to AMC, CAZ; Intermediate to DOX; Not determined: IPM"

To extract the individual resistances, the following steps are performed:

- The string is separated into sections by ";".
- Each section is stripped of whitespaces at the end or the beginning
- The resistance profile is determined by looking at the beginning of the section (Resistant => R, Susceptible => S, Intermediate => I, Not determined => ND)
- The antibiotics corresponding to the resistance profile are saved into a string.
- The antibiotic string is then split according to "," and stripped to remove whitespaces.
- A dictionary entryp is created for each antibiotic with the corresponding phenotype

The function then returns the final dictionary with a list of all antibiotics with corresponding phenotypes (R, S, I or ND)


```python
def parse_resistance_profile(resistance_string):
    """
    Parses resistance profile string into individual resistances.
    Returns a dictionary with antibiotic: phenotype (R, S, I or ND).
    """
    # initialize dictionary
    phenotypes = {}
        
    # Split input string by ";" to get resistance groups
    resistance_string = str(resistance_string).strip().rstrip(';') #remove ";" at the end and whitespaces
    sections = resistance_string.split(';') # split
    
    for section in sections:
        section = section.strip() # remove whitespace if present
            
        # Determine resistance profile
        if section.startswith('Resistant'):
            phenotype = 'R'
            antibiotics_str = section.replace('Resistant to', '').strip()
        elif section.startswith('Susceptible'):
            phenotype = 'S'
            antibiotics_str = section.replace('Susceptible to', '').strip()
        elif section.startswith('Intermediate'):
            phenotype = 'I'
            antibiotics_str = section.replace('Intermediate to', '').strip()
        elif section.startswith('Not determined'):
            phenotype = 'ND'
            antibiotics_str = section.replace('Not determined:', '').strip()
        
        # Split antibiotics by ","
        antibiotics = [ab.strip() for ab in antibiotics_str.split(',')]
        
        for antibiotic in antibiotics:
            if antibiotic:
                phenotypes[antibiotic] = phenotype
    
    return phenotypes
```

## **1. Load and Preprocess Data**
We will start by loading the CNV feature matrix and the sample metadata. We will then make sure that all samples in the feature matrix have a corresponding resistance profile for the antibiotic **SXT**.

### **1.1 Load CNV Feature Matrix**

```python
# Load raw CNV depth matrix
cnv_matrix = pd.read_csv('Data/cnv_depth_matrix.tsv', sep='\t', index_col=0)

print(f'Shape: {cnv_matrix.shape}')
print(f'Data types: {cnv_matrix.dtypes.unique()}')
print(f'Missing values: {cnv_matrix.isna().sum().sum()} / {cnv_matrix.size} ({100*cnv_matrix.isna().sum().sum()/cnv_matrix.size:.2f}%)')
```

### **1.2 Load Metadata**

```python
# Load metadata with resistance profiles
metadata = pd.read_csv('../genome_metadata.csv')

print(f'Shape: {metadata.shape}')
print(f'\nColumns: {metadata.columns}')
print(f'\n')
print(metadata.head())
```

### **1.3 Clean Metadata**
Let's keep only samples in the metadata that are also in the CNV feature matrix. 

```python
# Find common samples between CNV data and metadata and keep only common samples
metadata.set_index("sra_accession", inplace=True)

# Remove duplicate samples in both dataframes
metadata = metadata.loc[~metadata.index.duplicated()]
cnv_matrix = cnv_matrix.loc[~cnv_matrix.index.duplicated()]

# Find common samples
common_samples = cnv_matrix.index.intersection(metadata.index)

# Keep only samples in the metadata file that are in the cnv file
metadata = metadata.loc[common_samples]
cnv_matrix = cnv_matrix.loc[common_samples]

print(f"\nMerged data shape:")
print(f"CNVs: {cnv_matrix.shape}")
print(f"Metadata: {metadata.shape}")

# Save the cleaned file
metadata.to_csv("../genome_metadata_cleaned.csv")
```

### **1.4 Parse Resistance Profiles**
Now that we have loaded and cleaned the metadata, we will extract the resistance profile for each sample using the function parse_resistance_profile defined above.
We will apply the function to each row in the metadata to obtain a dictionary with {antibiotic1: profile1, antibotic2: profile2, etc.} for each sample.

We will then convert the dictionary to a data frame (each column is an antibiotic) and add the sample name as index.

```python
# Parse resistances for each sample: outputs a dictionary with antibiotic : profile pairs for each row (each sample)
parsed_resistances = metadata["resistance_profile"].apply(parse_resistance_profile) # apply the function to each row

# Convert to DataFrame (json_normalize creates one column per entry in the dictionary)
resistances_df = pd.json_normalize(parsed_resistances)

# Add sample name and set it as index
resistances_df['sra_accession'] = metadata.index.values
resistances_df = resistances_df.set_index('sra_accession')

print(f'Resistance profiles shape: {resistances_df.shape}')
print(f'Antibiotics detected: {resistances_df.columns.tolist()}\n')
print(resistances_df.head())
```

### **1.5 Filter by SXT Resistance Profile**
Finally, we will keep only samples that have a valid resistance profile for the antibiotic we are interested in (SXT).

```python
# Extract SXT resistance profile
SXT_profile = resistances_df['SXT']

# Keep only samples with valid SXT phenotype (S or R)
valid_samples = SXT_profile.isin(['S', 'R'])
cnv_matrix = cnv_matrix.loc[valid_samples]
SXT_profile = SXT_profile.loc[valid_samples]

# Convert to binary (0=Susceptible, 1=Resistant)
SXT_profile = (SXT_profile == 'R').astype(int)

print(f'Valid samples: {len(cnv_matrix)}')
print(f'\nSXT resistance distribution:')
print(SXT_profile.value_counts())
```

## **2. Preliminary data cleaning**
We will now perform some simple data cleaning to remove duplicated samples and features and fill empty values.

### **2.1 Remove duplicated samples**

```python
# Identify duplicate samples based on CNV values
duplicate_samples = cnv_matrix.duplicated()

print(f'Duplicate samples: {duplicate_samples.sum()}')

cnv_matrix = cnv_matrix.loc[~duplicate_samples]
SXT_profile = SXT_profile.loc[~duplicate_samples]
    
print(f'\nAfter removing duplicates:')
print(f'Shape: {cnv_matrix.shape}')
```

### **2.2 Remove duplicated features**

```python
# Find duplicate features
cnv_T = cnv_matrix.T
duplicate_features = cnv_T.duplicated()

print(f'Duplicate features: {duplicate_features.sum()}')
print(f'Before filtering: {cnv_matrix.shape}')

# Remove duplicates
cnv_matrix = cnv_matrix.loc[:, ~duplicate_features]

print(f'After filtering: {cnv_matrix.shape}')
print(f'Removed: {duplicate_features.sum()} duplicate features')
```

### **2.3 Fill NaN values**
Finally, we will fill empty values. The pipeline creates empty features by design: when a gene/promoter is absent for a sample, the CNV will not be calculated and will be left empty.

I want to replace these empty values with an arbitrarely low number (-20). Since the cnv_depth is already converted to log2 coverage, a very low value (-20) corresponds to a coverage close to 0. 

```python
print(f'NaN values before filling: {cnv_matrix.isna().sum().sum()} / {cnv_matrix.size}')

cnv_matrix.fillna(-20, inplace=True)

print(f'\nNaN values after filling: {cnv_matrix.isna().sum().sum()}')
print(f'\nFinal dataset shape: {cnv_matrix.shape}')
```

## **3. EDA and visualizations**
Let's now perform some simple data exploration to understand our data better. 

### **3.1 Summary statistics**
Let's calculate some summary statistics per sample:

- Mean CNV
- Median CNV
- Standard deviation of CNV
- Min CNV
- Max CNV
- Quantiles
- Interquantile range

```python
# Calculate per-sample statistics
sample_stats = pd.DataFrame({
    'sample': cnv_matrix.index,
    'mean': cnv_matrix.mean(axis=1).values,
    'median': cnv_matrix.median(axis=1).values,
    'std': cnv_matrix.std(axis=1).values,
    'min': cnv_matrix.min(axis=1).values,
    'max': cnv_matrix.max(axis=1).values,
    'q25': cnv_matrix.quantile(0.25, axis=1).values,
    'q75': cnv_matrix.quantile(0.75, axis=1).values
})

sample_stats['iqr'] = sample_stats['q75'] - sample_stats['q25']
sample_stats = sample_stats.set_index('sample')

print('Sample Statistics Summary:')
print(f'Mean CNV (across samples): {sample_stats["mean"].mean():.4f}')
print(f'Max value: {sample_stats["max"].max():.4f}')
print(f'Min value: {sample_stats["min"].min():.4f}')
print(f'Median std dev: {sample_stats["std"].mean():.4f}')
```

Let's also visualize some of these statistics

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Mean distribution
axes[0, 0].hist(sample_stats['median'], bins=50, alpha=0.7, edgecolor='black')
axes[0, 0].set_xlabel('Median CNV (log2 ratio)', fontweight='bold')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title('Distribution of Median CNVs', fontweight='bold')
axes[0, 0].grid(alpha=0.3)

# Standard deviation distribution
axes[0, 1].hist(sample_stats['std'], bins=50, alpha=0.7, edgecolor='black')
axes[0, 1].set_xlabel('CNV Standard Deviation', fontweight='bold')
axes[0, 1].set_ylabel('Frequency')
axes[0, 1].set_title('Distribution of CNV Std Dev', fontweight='bold')
axes[0, 1].grid(alpha=0.3)

# Max value distribution
axes[1, 0].hist(sample_stats['max'], bins=50, alpha=0.7, edgecolor='black')
axes[1, 0].set_xlabel('Max CNV (log2 ratio)', fontweight='bold')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title('Max CNV Per Sample', fontweight='bold')
axes[1, 0].grid(alpha=0.3)

# Min value distribution
axes[1, 1].hist(sample_stats['min'], bins=50, alpha=0.7, edgecolor='black')
axes[1, 1].set_xlabel('Min CNV (log2 ratio)', fontweight='bold')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].set_title('Min CNV Per Sample', fontweight='bold')
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('Figures/01_statistics.png', dpi=300, bbox_inches='tight')
plt.show()
```

We see that the median is centered around 0, which is what we would expect (log2(0)=1). Indeed, we would expect more or less the same number of genetic regions that have a copy number variation below 0 as above 0. 
Two samples have a very high standard deviation in comparison to the others, this is something that we will need to look at more in detail later on. 

Similarly, two samples have a very high max CNV (close 6, which would mean that the region is duplicated 64 times, which seems highly unlikely). We will need to adress this later on. 

The min CNV is -20 for all samples, which makes sense as we set all empty values to -20 in a previous step (these regions are not present in the genome). 

### **3.2 Boxplots of CNV values per sample**
Let's now have a look at the values for each individual sample using a boxplot)

```python
# Sort samples by mean CNV
sorted_samples = sample_stats.sort_values('mean', ascending=False).index
cnv_sorted = cnv_matrix.loc[sorted_samples]

fig, ax = plt.subplots(figsize=(16, 8))
sns.boxplot(cnv_sorted.T.values, showfliers=False)
ax.set_xlabel('Sample', fontweight='bold')
ax.set_ylabel('CNV log2 Ratio', fontweight='bold')
ax.set_title('CNV value distributions', fontweight='bold', fontsize=13)
plt.xticks(rotation=90, fontsize=8)
plt.tight_layout()
plt.savefig('Figures/01_boxplot_per_sample.png', dpi=300, bbox_inches='tight')
plt.show()
```

We immediately see that there are two samples that stand out by having very different distributions. These samples seem to have a lot of values to be close to 0, and at the mean time also have higher than average max values. We will need to filter out these samples before building our model. To avoid data leakage, we will take the decision concerning which samples to remove from the analysis **after splitting the data**.

## **4. Save preprocessed data**
Let's save the preprocessed data.

```python
# Save cleaned CNV matrix
cnv_matrix.to_csv('Data/cnvs_cleaned.csv')

# Save SXT resistance profile
SXT_profile.to_csv('Data/SXT_cleaned.csv', header=['SXT_binary'])

print(f'Shape: {cnv_matrix.shape}')
```
