```python
import numpy as np, pandas as pd

cnv = pd.read_csv("Data/cnv_depth_matrix.tsv", sep="\t", index_col=0)
pres = pd.DataFrame(
    np.select([cnv < -1.0, cnv >= 0.5], [0, 2], default=1),
    index=cnv.index, columns=cnv.columns
).where(cnv.notna())
pres.to_csv("Data/cnvs_gene_presence.tsv", sep="\t")
```

```python
for f in ["cnv_depth__IBNLCO_00792__bpeB__gene__promoter",
          "cnv_depth__IBNLCO_02855__secF__gene__promoter"]:
    print(f.split("__")[2], pres[f].value_counts().to_dict())
```
